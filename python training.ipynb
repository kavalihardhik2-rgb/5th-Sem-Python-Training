```python
# Program to check whether a number is a Strong Number

num = int(input("Enter a number: "))

temp = num
sum = 0

while temp > 0:
    digit = temp % 10

    # Find factorial of the digit
    fact = 1
    for i in range(1, digit + 1):
        fact = fact * i

    sum = sum + fact
    temp = temp // 10

if sum == num:
    print(num, "is a Strong Number")
else:
    print(num, "is Not a Strong Number")
```

```python
# Program to check whether a number is a palindrome

num = int(input("Enter a number: "))

original = num
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

if original == reverse:
    print(original, "is a Palindrome Number")
else:
    print(original, "is Not a Palindrome Number")
```

```python
# Write a Python program to find the second largest and third smallest
# elements in a list without using sort()

n = int(input("Enter the number of elements: "))

lst = []

for i in range(n):
    element = int(input("Enter element: "))
    lst.append(element)

# Find second largest
largest = second_largest = float('-inf')

for num in lst:
    if num > largest:
        second_largest = largest
        largest = num
    elif num > second_largest and num != largest:
        second_largest = num

# Find third smallest
smallest = second_smallest = third_smallest = float('inf')

for num in lst:
    if num < smallest:
        third_smallest = second_smallest
        second_smallest = smallest
        smallest = num
    elif num < second_smallest and num != smallest:
        third_smallest = second_smallest
        second_smallest = num
    elif num < third_smallest and num != smallest and num != second_smallest:
        third_smallest = num

print("Second Largest =", second_largest)
print("Third Smallest =", third_smallest)
```

```python
# Program to merge two lists, separate even and odd numbers,
# and display both lists in ascending order

n1 = int(input("Enter the number of elements in List 1: "))
list1 = []

for i in range(n1):
    element = int(input("Enter element: "))
    list1.append(element)

n2 = int(input("Enter the number of elements in List 2: "))
list2 = []

for i in range(n2):
    element = int(input("Enter element: "))
    list2.append(element)

# Merge the two lists
merged = list1 + list2

# Separate even and odd numbers
even = []
odd = []

for num in merged:
    if num % 2 == 0:
        even.append(num)
    else:
        odd.append(num)

# Sort both lists in ascending order
even.sort()
odd.sort()

# Display the results
print("\nMerged List:", merged)
print("Even Numbers (Ascending):", even)
print("Odd Numbers (Ascending):", odd)
```

```python
#write a python program to count the frequency of each word in a paragraph using a dictonary
#display the words in descending order of their frequency

paragraph = input("Enter a paragraph: ")

# Convert to lowercase and split into words
words = paragraph.lower().split()

frequency = {}

# Count the frequency of each word
for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

# Sort the dictionary by frequency in descending order
sorted_words = sorted(frequency.items(), key=lambda x: x[1], reverse=True)

print("\nWord Frequencies (Descending Order):")

for word, count in sorted_words:
    print(word, ":", count)
```

```python
#Library management system 
# create a book class with attributes title, author, and price.
# create a derived class EBook that adds file_size. Keep price private and provide getter/setter methods.
# override a display_details in EBook.
# concepts: class, constructor, encapsulation, inheritence, polymorphism

class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.__price = price      

    # Getter
    def get_price(self):
        return self.__price

    # Setter
    def set_price(self, price):
        self.__price = price

    # Method to display details
    def display_details(self):
        print("\nBook Details")
        print("Title :", self.title)
        print("Author:", self.author)
        print("Price :", self.get_price())


# Derived Class
class EBook(Book):
    def __init__(self, title, author, price, file_size):
        super().__init__(title, author, price)
        self.file_size = file_size

    # Method Overriding (Polymorphism)
    def display_details(self):
        print("\nEBook Details")
        print("Title      :", self.title)
        print("Author     :", self.author)
        print("Price      :", self.get_price())
        print("File Size  :", self.file_size, "MB")


# Main Program
title = input("Enter Book Title: ")
author = input("Enter Author Name: ")
price = float(input("Enter Book Price: "))
file_size = float(input("Enter File Size (MB): "))

ebook = EBook(title, author, price, file_size)

print("\nUsing Getter Method")
print("Price:", ebook.get_price())

new_price = float(input("\nEnter New Price: "))
ebook.set_price(new_price)

print("\nUpdated Details")
ebook.display_details()
```

```python
#Vehicle rental system
#create a base class class vehicle with brand and rent_per_day
#create derived classes car and bike. implement a method calculate_rent(days) in each class(different pricing
#rules) rental deatils
#concepts: inheritence, method overriding, polymorphism

class Vehicle:
    def __init__(self, brand, rent_per_day):
        self.brand = brand
        self.rent_per_day = rent_per_day

    def calculate_rent(self, days):
        return self.rent_per_day * days

    def rental_details(self, days):
        print("\nBrand:", self.brand)
        print("Rent Per Day:", self.rent_per_day)
        print("Days:", days)
        print("Total Rent:", self.calculate_rent(days))


# Derived Class - Car
class Car(Vehicle):
    # Method Overriding
    def calculate_rent(self, days):
        return (self.rent_per_day * days) + 500   


# Derived Class - Bike
class Bike(Vehicle):
    # Method Overriding
    def calculate_rent(self, days):
        return self.rent_per_day * days           


# Main Program
print("1. Car")
print("2. Bike")

choice = int(input("Enter your choice: "))

brand = input("Enter Vehicle Brand: ")
rent = float(input("Enter Rent Per Day: "))
days = int(input("Enter Number of Days: "))

if choice == 1:
    vehicle = Car(brand, rent)
elif choice == 2:
    vehicle = Bike(brand, rent)
else:
    print("Invalid Choice")
    exit()


vehicle.rental_details(days)
```

```python
#Online shopping kart
#create a product class with private attribute name and price
#add methods to update the price and display product details. create a discountedProduct class that applies a percentage discount.
#calculate the final bill
#Concepts: encapsulation, inheritance, polymorphism

class Product:
    def __init__(self, name, price):
        self.__name = name      # Private attribute
        self.__price = price    # Private attribute

    # Getter methods
    def get_name(self):
        return self.__name

    def get_price(self):
        return self.__price

    # Setter method
    def set_price(self, price):
        self.__price = price

    # Display product details
    def display_details(self):
        print("\nProduct Name :", self.get_name())
        print("Price        :", self.get_price())

    # Calculate final bill
    def final_bill(self):
        return self.get_price()


# Derived Class
class DiscountedProduct(Product):
    def __init__(self, name, price, discount):
        super().__init__(name, price)
        self.discount = discount

    # Method Overriding (Polymorphism)
    def final_bill(self):
        return self.get_price() - (self.get_price() * self.discount / 100)

    # Overriding display method
    def display_details(self):
        print("\nProduct Name :", self.get_name())
        print("Original Price :", self.get_price())
        print("Discount :", self.discount, "%")
        print("Final Bill :", self.final_bill())


# Main Program
name = input("Enter Product Name: ")
price = float(input("Enter Product Price: "))
discount = float(input("Enter Discount Percentage: "))

product = DiscountedProduct(name, price, discount)

print("\nProduct Details")
product.display_details()

# Update Price
new_price = float(input("\nEnter New Price: "))
product.set_price(new_price)

print("\nUpdated Product Details")
product.display_details()
```

```python
#school management system
#create a base class person
#derive student and teacher classes. both should override a show_deatils() method. create objects of
#  both classes and display their deatils using a loop
#concepts: inheritence, method overriding, polymorphism

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def show_details(self):
        print("Name:", self.name)
        print("Age:", self.age)
# Derived Class - Student
class Student(Person):
    def __init__(self, name, age, roll_no):
        super().__init__(name, age)
        self.roll_no = roll_no
    # Method Overriding
    def show_details(self):
        print("\nStudent Details")
        print("Name:", self.name)
        print("Age:", self.age)
        print("Roll Number:", self.roll_no)
# Derived Class - Teacher
class Teacher(Person):
    def __init__(self, name, age, subject):
        super().__init__(name, age)
        self.subject = subject
    # Method Overriding
    def show_details(self):
        print("\nTeacher Details")
        print("Name:", self.name)
        print("Age:", self.age)
        print("Subject:", self.subject)
# Main Program
student = Student("Rahul", 20, 101)
teacher = Teacher("Anita", 35, "Mathematics")
# List of objects
people = [student, teacher]
# Polymorphism using loop
for person in people:
    person.show_details()
```

```python
def linear_search(arr, key):
    for i in range(len(arr)):
        if arr[i] == key:
            return i
    return -1

arr = [5, 12, 18, 25, 40]

key = int(input("Enter element to search: "))

result = linear_search(arr, key)

if result != -1:
    print("Element found at index", result)
else:
    print("Element not found")
```

```python
def binary_search(arr, key):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == key:
            return mid
        elif key < arr[mid]:
            high = mid - 1
        else:
            low = mid + 1

    return -1

arr = [10, 20, 30, 40, 50, 60, 70, 80]

key = int(input("Enter element to search: "))

result = binary_search(arr, key)

if result != -1:
    print("Element found at index", result)
else:
    print("Element not found")
```

```python
#BUBBLE SORT

arr = list(map(int, input("Enter the elements: ").split()))
n = len(arr)

for i in range(n - 1):
    swapped = False

    for j in range(n - 1 - i):
        if arr[j] > arr[j + 1]:
            arr[j], arr[j + 1] = arr[j + 1], arr[j]
            swapped = True

    # If no swapping occurred, the array is already sorted
    if not swapped:
        break

print("Sorted array:", arr)
```

```python
# Selection Sort

arr = list(map(int, input("Enter the elements: ").split()))
n = len(arr)

for i in range(n - 1):
    min_index = i

    for j in range(i + 1, n):
        if arr[j] < arr[min_index]:
            min_index = j

    
    arr[i], arr[min_index] = arr[min_index], arr[i]

print("Sorted array:", arr)
```

```python
# Insertion Sort

arr = list(map(int, input("Enter the elements: ").split()))
n = len(arr)

for i in range(1, n):
    key = arr[i]
    j = i - 1

    while j >= 0 and arr[j] > key:
        arr[j + 1] = arr[j]
        j -= 1

    arr[j + 1] = key

print("Sorted array:", arr)
```

```python
# Merge Sort

def merge_sort(arr):
    if len(arr) > 1:
        mid = len(arr) // 2

        left = arr[:mid]
        right = arr[mid:]

        # Recursively sort both halves
        merge_sort(left)
        merge_sort(right)

        i = j = k = 0

        # Merge the sorted halves
        while i < len(left) and j < len(right):
            if left[i] < right[j]:
                arr[k] = left[i]
                i += 1
            else:
                arr[k] = right[j]
                j += 1
            k += 1

        # Copy remaining elements of left half
        while i < len(left):
            arr[k] = left[i]
            i += 1
            k += 1

        # Copy remaining elements of right half
        while j < len(right):
            arr[k] = right[j]
            j += 1
            k += 1

# Main Program
arr = list(map(int, input("Enter the elements: ").split()))

merge_sort(arr)

print("Sorted array:", arr)
```

```python
# Node class
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


# Linked List class
class LinkedList:
    def __init__(self):
        self.head = None

    # Insert at end
    def insert(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        temp = self.head
        while temp.next:
            temp = temp.next
        temp.next = new_node

    # Delete at beginning
    def delete_beginning(self):
        if self.head is None:
            print("Linked List is empty")
            return

        self.head = self.head.next

    # Delete in middle (delete by value)
    def delete_middle(self, value):
        if self.head is None:
            print("Linked List is empty")
            return

        # If first node contains the value
        if self.head.data == value:
            self.head = self.head.next
            return

        prev = None
        temp = self.head

        while temp and temp.data != value:
            prev = temp
            temp = temp.next

        if temp is None:
            print("Value not found")
            return

        prev.next = temp.next

    # Delete at end
    def delete_end(self):
        if self.head is None:
            print("Linked List is empty")
            return

        if self.head.next is None:
            self.head = None
            return

        temp = self.head

        while temp.next.next:
            temp = temp.next

        temp.next = None

    # Display Linked List
    def display(self):
        if self.head is None:
            print("Linked List is empty")
            return

        temp = self.head
        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next
        print("None")


# Driver Code
ll = LinkedList()

ll.insert(10)
ll.insert(20)
ll.insert(30)
ll.insert(40)
ll.insert(50)

print("Original Linked List:")
ll.display()

# Delete at beginning
ll.delete_beginning()
print("\nAfter deleting at beginning:")
ll.display()

# Delete middle node (value = 30)
ll.delete_middle(30)
print("\nAfter deleting middle node (30):")
ll.display()

# Delete at end
ll.delete_end()
print("\nAfter deleting at end:")
ll.display()
```

```python
'''Question 1
Write a Python program to reverse a queue without using the built-in reverse() method.
Example:
Input:
10 20 30 40 50

Output:
50 40 30 20 10'''
from collections import deque

queue = deque([10,20,30,40,50])

stack = []
while queue:
    stack.append(queue.popleft())

while stack:
    queue.append(stack.pop())

print("Reversed Queue:", *queue)
```

```python
'''Question 2
Write a Python program to find the maximum and minimum elements in a queue without changing the original queue.'''

from collections import deque

queue = deque([10, 25, 5, 40, 15])

# Copy the queue
temp = deque(queue)

maximum = temp[0]
minimum = temp[0]

while temp:
    value = temp.popleft()

    if value > maximum:
        maximum = value

    if value < minimum:
        minimum = value

print("Maximum:", maximum)
print("Minimum:", minimum)

print("Original Queue:", *queue)

```

```python
'''Question 3
Write a Python program to remove all duplicate elements from a queue while preserving their original order.
Example:
Input:
10 20 20 30 10 40
Output:
10 20 30 40'''

from collections import deque

queue = deque([10, 20, 20, 30, 10, 40])
result = deque()

while queue:
    value = queue.popleft()

    if value not in result:
        result.append(value)

print("Queue after removing duplicates:", *result)

```

```python
'''Question 6
Write a Python program to compare two queues and determine whether they contain the same elements in the same order.
Example:
Queue 1:
10 20 30

Queue 2:
10 20 30

Output:
Queues are Equal'''
from collections import deque

queue1 = deque([10, 20, 30])
queue2 = deque([10, 20, 30])

if len(queue1) != len(queue2):
    print("Queues are Not Equal")
else:
    equal = True

    while queue1:
        a = queue1.popleft()
        b = queue2.popleft()

        if a != b:
            equal = False
            break

    if equal:
        print("Queues are Equal")
    else:
        print("Queues are Not Equal")
```

```python
'''Question 9
Write a Python program to move all elements smaller than a given value to a separate queue, while keeping the remaining elements in the original queue.
Example:
Queue:
10 25 15 40 20 50

Value = 25

New Queue:
10 15 20'''    
from collections import deque

queue = deque([10, 25, 15, 40, 20, 50])

value = 25

new_queue = deque()
remaining_queue = deque()

while queue:
    element = queue.popleft()

    if element < value:
        new_queue.append(element)
    else:
        remaining_queue.append(element)

queue = remaining_queue

print("New Queue:", *new_queue)
print("Original Queue:", *queue)
```

```python
# Type: Level Order Traversal
# Technique: Breadth First Search (BFS)
# Data Structure Used: Queue

from collections import deque


class Node:

    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


# Create Tree

root = Node(10)

root.left = Node(20)
root.right = Node(30)

root.left.left = Node(40)
root.left.right = Node(50)

root.right.right = Node(60)


# Level Order Traversal

def level_order(root):

    if root is None:
        return

    queue = deque()

    queue.append(root)

    while queue:

        current = queue.popleft()

        print(current.data, end=" ")

        if current.left:
            queue.append(current.left)

        if current.right:
            queue.append(current.right)


print("Level Order Traversal:")
level_order(root)
```

```python
# Type: Binary Search Tree (BST)
# Operation: Search

class Node:

    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def insert(root, data):

    if root is None:
        return Node(data)

    if data < root.data:
        root.left = insert(root.left, data)

    elif data > root.data:
        root.right = insert(root.right, data)

    return root


def search(root, target):

    if root is None:
        return False

    if root.data == target:
        return True

    if target < root.data:
        return search(root.left, target)

    return search(root.right, target)


# Create BST

root = None

root = insert(root, 50)
root = insert(root, 30)
root = insert(root, 70)
root = insert(root, 20)
root = insert(root, 40)
root = insert(root, 60)
root = insert(root, 80)


# Search Element

target = 60

if search(root, target):
    print("Element Found")
else:
    print("Element Not Found")
```

```python
# Question 4
# Type: Height of Binary Tree

class Node:

    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def height(root):

    if root is None:
        return -1

    left_height = height(root.left)
    right_height = height(root.right)

    return 1 + max(left_height, right_height)


root = Node(10)

root.left = Node(20)
root.right = Node(30)

root.left.left = Node(40)
root.left.right = Node(50)

root.left.right.left = Node(60)

print("Height =", height(root))
```

```python
# Question 5
# Type: Count Leaf Nodes

class Node:

    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def count_leaf(root):

    if root is None:
        return 0

    if root.left is None and root.right is None:
        return 1

    return count_leaf(root.left) + count_leaf(root.right)


root = Node(10)

root.left = Node(20)
root.right = Node(30)

root.left.left = Node(40)
root.left.right = Node(50)

root.right.right = Node(60)

print("Leaf Nodes =", count_leaf(root))
```

```python
# Question 6
# Type: Count Total Nodes

class Node:

    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def count_nodes(root):

    if root is None:
        return 0

    return 1 + count_nodes(root.left) + count_nodes(root.right)


root = Node(10)

root.left = Node(20)
root.right = Node(30)

root.left.left = Node(40)
root.left.right = Node(50)

print("Total Nodes =", count_nodes(root))
```

```python
# Question 7
# Type: BST Search

class Node:

    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def search(root, target):

    if root is None:
        return False

    if root.data == target:
        return True

    if target < root.data:
        return search(root.left, target)

    return search(root.right, target)


root = Node(50)

root.left = Node(30)
root.right = Node(70)

root.left.left = Node(20)
root.left.right = Node(40)

root.right.left = Node(60)
root.right.right = Node(80)

target = 60

if search(root, target):
    print("Element Found")
else:
    print("Element Not Found")
```

```python
# Question 8
# Type: Minimum and Maximum in BST

class Node:

    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def find_min(root):

    while root.left is not None:
        root = root.left

    return root.data


def find_max(root):

    while root.right is not None:
        root = root.right

    return root.data


root = Node(50)

root.left = Node(30)
root.right = Node(70)

root.left.left = Node(20)
root.left.right = Node(40)

root.right.left = Node(60)
root.right.right = Node(80)

print("Minimum =", find_min(root))
print("Maximum =", find_max(root))
```

```python
# Question 9
# Type: Validate Binary Search Tree

class Node:

    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def is_bst(root, minimum, maximum):

    if root is None:
        return True

    if root.data <= minimum or root.data >= maximum:
        return False

    return (
        is_bst(root.left, minimum, root.data)
        and
        is_bst(root.right, root.data, maximum)
    )


root = Node(50)

root.left = Node(30)
root.right = Node(70)

root.left.left = Node(20)
root.left.right = Node(40)

root.right.left = Node(60)
root.right.right = Node(80)

if is_bst(root, float("-inf"), float("inf")):
    print("Valid BST")
else:
    print("Invalid BST")
```

```python
# Question 10
# Type: Kth Smallest Element in BST

class Node:

    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def kth_smallest(root, k):

    values = []

    def inorder(node):

        if node is None:
            return

        inorder(node.left)
        values.append(node.data)
        inorder(node.right)

    inorder(root)

    return values[k - 1]


root = Node(50)

root.left = Node(30)
root.right = Node(70)

root.left.left = Node(20)
root.left.right = Node(40)

root.right.left = Node(60)
root.right.right = Node(80)

k = 3

print("Kth Smallest =", kth_smallest(root, k))
```

```python
# Question 11
# Type: Level with Maximum Nodes
# Technique: BFS

from collections import deque


class Node:

    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def max_node_level(root):

    if root is None:
        return -1, 0

    queue = deque([root])
    level = 0

    max_level = 0
    max_nodes = 0

    while queue:

        count = len(queue)

        if count > max_nodes:
            max_nodes = count
            max_level = level

        for _ in range(count):

            current = queue.popleft()

            if current.left:
                queue.append(current.left)

            if current.right:
                queue.append(current.right)

        level += 1

    return max_level, max_nodes


root = Node(10)

root.left = Node(20)
root.right = Node(30)

root.left.left = Node(40)
root.left.right = Node(50)

root.right.left = Node(60)
root.right.right = Node(70)

level, nodes = max_node_level(root)

print("Level =", level)
print("Nodes =", nodes)
```

```python
# Question 12
# Type: Mirror of Binary Tree

class Node:

    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def mirror(root):

    if root is None:
        return

    root.left, root.right = root.right, root.left

    mirror(root.left)
    mirror(root.right)


def inorder(root):

    if root is None:
        return

    inorder(root.left)
    print(root.data, end=" ")
    inorder(root.right)


root = Node(10)

root.left = Node(20)
root.right = Node(30)

root.left.left = Node(40)
root.left.right = Node(50)

print("Original Inorder:")
inorder(root)

mirror(root)

print("\nMirror Inorder:")
inorder(root)
```

```python
# Question 13
# Type: Sum of Nodes at a Given Level
# Technique: BFS

from collections import deque


class Node:

    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def level_sum(root, target_level):

    if root is None:
        return 0

    queue = deque([root])
    level = 0

    while queue:

        count = len(queue)
        total = 0

        for _ in range(count):

            current = queue.popleft()

            if level == target_level:
                total += current.data

            if current.left:
                queue.append(current.left)

            if current.right:
                queue.append(current.right)

        if level == target_level:
            return total

        level += 1

    return 0


root = Node(10)

root.left = Node(20)
root.right = Node(30)

root.left.left = Node(40)
root.left.right = Node(50)

root.right.left = Node(60)
root.right.right = Node(70)

level = 2

print("Sum =", level_sum(root, level))
```

```python
# Question 14
# Type: Lowest Common Ancestor in BST

class Node:

    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def lca(root, value1, value2):

    if root is None:
        return None

    if value1 < root.data and value2 < root.data:
        return lca(root.left, value1, value2)

    if value1 > root.data and value2 > root.data:
        return lca(root.right, value1, value2)

    return root


root = Node(50)

root.left = Node(30)
root.right = Node(70)

root.left.left = Node(20)
root.left.right = Node(40)

root.right.left = Node(60)
root.right.right = Node(80)

value1 = 20
value2 = 40

result = lca(root, value1, value2)

print("LCA =", result.data)
```

```python
# Question 15
# Type: BST Deletion

class Node:

    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def insert(root, data):

    if root is None:
        return Node(data)

    if data < root.data:
        root.left = insert(root.left, data)

    elif data > root.data:
        root.right = insert(root.right, data)

    return root


def find_min(root):

    while root.left:
        root = root.left

    return root


def delete(root, target):

    if root is None:
        return None

    if target < root.data:
        root.left = delete(root.left, target)

    elif target > root.data:
        root.right = delete(root.right, target)

    else:

        # Case 1: No child

        if root.left is None and root.right is None:
            return None

        # Case 2: Only right child

        if root.left is None:
            return root.right

        # Case 2: Only left child

        if root.right is None:
            return root.left

        # Case 3: Two children

        successor = find_min(root.right)

        root.data = successor.data

        root.right = delete(root.right, successor.data)

    return root


def inorder(root):

    if root is None:
        return

    inorder(root.left)
    print(root.data, end=" ")
    inorder(root.right)


# Create BST

root = None

root = insert(root, 50)
root = insert(root, 30)
root = insert(root, 70)
root = insert(root, 20)
root = insert(root, 40)
root = insert(root, 60)
root = insert(root, 80)


print("Before Deletion:")
inorder(root)

root = delete(root, 30)

print("\nAfter Deletion:")
inorder(root)
```

```python
'''Question 16 – Sum of Nodes at Each Level
Write a Python program to calculate and display the sum of all nodes present at each level of a binary tree.

Example:

          10
        /    \
       20     30
      /  \   /  \
     40   50 60   70
Output:

Level 0 → 10 Level 1 → 50 Level 2 → 220'''    
from collections import deque

class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def sum_at_each_level(root):
    if root is None:
        return

    queue = deque([root])
    level = 0

    while queue:
        level_sum = 0
        level_size = len(queue)

        # Process all nodes at the current level
        for _ in range(level_size):
            node = queue.popleft()
            level_sum += node.data

            if node.left:
                queue.append(node.left)

            if node.right:
                queue.append(node.right)

        print(f"Level {level} → {level_sum}")
        level += 1


# Creating the binary tree
root = Node(10)

root.left = Node(20)
root.right = Node(30)

root.left.left = Node(40)
root.left.right = Node(50)

root.right.left = Node(60)
root.right.right = Node(70)

# Calculate sum at each level
sum_at_each_level(root)
```

```python
'''Question 17 – Check Height Balance
Write a Python program to determine whether a binary tree is height-balanced.

A binary tree is considered balanced if, for every node, the difference between the height of its left subtree and right subtree is not greater than 1.

Example:

          10
        /    \
       20     30
      /  \
     40   50
Output:

Balanced Tree'''   
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def check_balanced(root):

    def height(node):
        if node is None:
            return 0

        left_height = height(node.left)

        # If left subtree is unbalanced
        if left_height == -1:
            return -1

        right_height = height(node.right)

        # If right subtree is unbalanced
        if right_height == -1:
            return -1

        # Check height difference
        if abs(left_height - right_height) > 1:
            return -1

        # Return height of current node
        return 1 + max(left_height, right_height)

    return height(root) != -1


# Creating the binary tree
root = Node(10)

root.left = Node(20)
root.right = Node(30)

root.left.left = Node(40)
root.left.right = Node(50)


# Check whether tree is balanced
if check_balanced(root):
    print("Balanced Tree")
else:
    print("Not a Balanced Tree")
```

```python
'''Question 18 – Find the Diameter of a Binary Tree
Write a Python program to find the diameter of a binary tree.

The diameter is the number of edges in the longest path between any two nodes in the tree. The path does not necessarily have to pass through the root.

Example:

          10
        /    \
       20     30
      /  \
     40   50
    /
   60
Output:

Diameter = 4'''    
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def diameter(root):
    max_diameter = [0]

    def height(node):
        if node is None:
            return 0

        left_height = height(node.left)
        right_height = height(node.right)

        # Diameter passing through current node
        current_diameter = left_height + right_height

        max_diameter[0] = max(max_diameter[0], current_diameter)

        return 1 + max(left_height, right_height)

    height(root)
    return max_diameter[0]


# Creating the binary tree
root = Node(10)

root.left = Node(20)
root.right = Node(30)

root.left.left = Node(40)
root.left.right = Node(50)

root.left.left.left = Node(60)


print("Diameter =", diameter(root))
```

```python
'''Question 19 – Find the Lowest Common Ancestor
Given a Binary Search Tree and two node values, write a Python program to find their Lowest Common Ancestor (LCA).

The Lowest Common Ancestor is the lowest node in the tree that has both given nodes as descendants.

Example:

          50
        /    \
       30     70
      /  \   /  \
     20  40 60   80
Input:

Node 1 = 20 Node 2 = 40

Output:

Lowest Common Ancestor = 30'''    
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def find_lca(root, n1, n2):
    current = root

    while current:
        if n1 < current.data and n2 < current.data:
            current = current.left

        elif n1 > current.data and n2 > current.data:
            current = current.right

        else:
            return current.data

    return None


# Creating the BST
root = Node(50)

root.left = Node(30)
root.right = Node(70)

root.left.left = Node(20)
root.left.right = Node(40)

root.right.left = Node(60)
root.right.right = Node(80)


n1 = 20
n2 = 40

result = find_lca(root, n1, n2)

print("Lowest Common Ancestor =", result)
```

```python
'''Question 20 – Find the Kth Largest Element in BST
Given a Binary Search Tree and an integer K, write a Python program to find the Kth largest element without sorting the elements separately.

Example:

          50
        /    \
       30     70
      /  \   /  \
     20  40 60   80
K = 3

Output:

3rd Largest Element = 60'''    

class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def kth_largest(root, k):
    count = 0
    result = None

    def reverse_inorder(node):
        nonlocal count, result

        if node is None or count >= k:
            return

        # Visit right subtree first
        reverse_inorder(node.right)

        count += 1

        if count == k:
            result = node.data
            return

        # Visit left subtree
        reverse_inorder(node.left)

    reverse_inorder(root)

    return result


# Creating the BST
root = Node(50)

root.left = Node(30)
root.right = Node(70)

root.left.left = Node(20)
root.left.right = Node(40)

root.right.left = Node(60)
root.right.right = Node(80)


k = 3

result = kth_largest(root, k)

print(f"{k}rd Largest Element =", result)
```
